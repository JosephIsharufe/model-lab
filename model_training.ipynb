```python
from pathlib import Path

PROJET_DIR = Path("/content/drive/MyDrive/model-training")

CHECKPONT_DIR = PROJET_DIR/"checkpoints"
CHECKPONT_DIR.mkdir(exist_ok=True, parents=True)
```

```python
checkpoint_path = CHECKPONT_DIR/"test_checkpoint.pt"

training_state = {"epoch": 1,
                  "model_name": "model-uno",
                  "completed_steps": 100}


torch.save(training_state,checkpoint_path)
print("Saved:", checkpoint_path)

restored_state = torch.load(
    checkpoint_path,
    map_location = "cpu",
    weights_only= True
)

print(restored_state)
```
